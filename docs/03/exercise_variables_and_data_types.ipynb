### 演習5
`insert()` と `remove()` を用いてリスト `[10, 20, 30, 40, 50]` を1つ右に回転させよ．すなわち，出力例のように最後の要素が最初に移動する．

**出力例**
```python
[50, 10, 20, 30, 40]
```

````python
# 解答用セル

a=[10, 20, 30, 40, 50]
last=a[-1]
a.remove(last)
a.insert(0, last)
print(a)
````
